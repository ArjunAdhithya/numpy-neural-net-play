## On a Neural Net

```python
W = [
    [0.5, -1.0, 2.0],
    [0.5, -1.0, 2.0],
    [0.5, -1.0, 2.0],    
]
b = [0.25, 0.25, 0.25]
```

```python
type(W), type(b)
```

```python
layer1 = multipleNeuron(n=3, n_inputs=3, activation=sigmoid, W=W, b=b)
layer2 = multipleNeuron(n=3, n_inputs=3, activation=sigmoid, W=W, b=b)
```

```python
neural_net = stackLayers([layer1, layer2])
forward_pass1 = neural_net.forward([1.0, 2.0, -1.0])
```

```python
forward_pass1
```

```python
target = [1,0,0]

mean_squared_error(target,forward_pass1)
```
